1. Write a ``Particle`` class that can be used to represent a particle with a mass, a 3-d position, and a 3-d velocity.

2. Write a method that can be used to compute the kinetic energy of the particle

3. Write a method that takes another particle as an argument and finds the distance between the two particles

5. Write a ``ChargedParticle`` class that inherits from the ``Particle`` class, but also has an attribute for the charge of the particle.


```python
import numpy as np

class Particle(object):

    def __init__(self, m, x, y, z, vx, vy, vz):
        self.m = m
        self.x = x
        self.y = y
        self.z = z
        self.vx = vx
        self.vy = vy
        self.vz = vz

    def speed(self):
        s = np.sqrt(self.vx**2 + self.vy**2 + self.vz**2)
        return s
        
    def findKE(self):
        KE = 1/2 * self.m * self.speed()**2
        return KE

    def find_distance(self, other):
        dx = abs(self.x - other.x)
        dy = abs(self.y - other.y)
        dz = abs(self.z - other.z)
        d = np.sqrt(dx**2 + dy**2 + dz**2)
        return d

    def change(self, dt):
        self.x = self.x + self.vx * dt
        self.y = self.y + self.vy * dt
        self.z = self.z + self.vz * dt    

class ChargedParticle(Particle):
    
    def __init__(self, particle, charge):
    
        super().__init__(particle.m, particle.x, particle.y, particle.z, particle.vx, particle.vy, particle.vz)
        # Add a new attribute for electric charge
        self.charge = charge
```

```python
p1 = Particle(5, 2, 2, 2, 1, 0, 1)
p2 = Particle(1, 0, 0, 0, 5, 5, 5)
print(p1.speed())
print(p2.speed())
print(p1.findKE())
print(p2.findKE())
print(p1.find_distance(p2))
p1.change(1)
p2.change(1)
print(p1.find_distance(p2))
newp1 = ChargedParticle(p1, 5)
print(newp1.charge)
```

```python
class ChargedParticle(Particle):
    
    def __init__(self, particle, charge):
    
        super().__init__(particle.m, particle.x, particle.y, particle.z, particle.vx, particle.vy, particle.vz)
        # Add a new attribute for electric charge
        self.charge = charge

    def interaction(self, other):
        interaction = False
        if self.find_distance(other) < 0.25:
            interaction = True
        if interaction == True:
            print('interaction')
            if self.charge == -1 * other.charge:
                print('merge')
                self.vx = 0
                self.vy = 0
                self.vz = 0
                self.charge = 0
                other.vx = 0
                other.vy = 0
                other.vz = 0
                other.charge = 0
            else:
                print('repel')
                self.vy = -1 * self.vx * (self.charge * other.charge) * (self.m / other.m)
                self.vy = -1 * self.vy * (self.charge * other.charge) * (self.m / other.m)
                self.vz = -1 * self.vz * (self.charge * other.charge) * (self.m / other.m)
                
                other.vx = -1 * other.vx * (self.charge * other.charge) * (other.m / self.m)
                other.vy = -1 * other.vy * (self.charge * other.charge) * (other.m / self.m)
                other.vz = -1 * other.vz * (self.charge * other.charge) * (other.m / self.m)
        else:
            print('no interaction')
```
